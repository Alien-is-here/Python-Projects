Face Detection Code

```python

import cv2
from pathlib import Path

site_packages = Path(cv2.__file__).parent.parent
cascade_path = site_packages / "~v2" / "data" / "haarcascade_frontalface_default.xml"       # finds the Haar face-detection XML.

face_cascade = cv2.CascadeClassifier(str(cascade_path))     # loads the face detector.

video_cap = cv2.VideoCapture(0)

while True:
    ret, frame = video_cap.read()

    if not ret:
        print("Failed to read camera")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)          # Converting image to grayscale

    faces = face_cascade.detectMultiScale(      # detects faces.
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    for (x, y, w, h) in faces:
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (255, 0, 0),
            2
        )

    cv2.imshow("Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("a"):
        break

video_cap.release()
cv2.destroyAllWindows()
```
